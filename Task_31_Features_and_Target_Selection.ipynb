# Task 31: Features and Target Selection\n**Veda Technology - AI & ML Internship**\n\nObjective: Select input features (X) and a target variable (y) using the Iris dataset.

```python
import pandas as pd
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()

# Create a DataFrame and add the flower species as the target column
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["species"] = pd.Series(iris.target).map(dict(enumerate(iris.target_names)))

print("Dataset preview:")
display(df.head())
print("Dataset shape:", df.shape)

# Select input features (X)
feature_columns = [
    "sepal length (cm)",
    "sepal width (cm)",
    "petal length (cm)",
    "petal width (cm)"
]
X = df[feature_columns].copy()

# Select target variable (y)
y = df["species"].copy()

print("\nSelected features (X):")
display(X.head())
print("\nSelected target (y):")
display(y.head())

print("\nX shape:", X.shape)
print("y shape:", y.shape)
print("\nTarget categories:")
print(y.value_counts())

# Validate the selection
assert "species" not in X.columns
assert len(X) == len(y)
assert X.isnull().sum().sum() == 0
assert y.isnull().sum() == 0

print("\nValidation successful! Task 31 completed.")
```

## Explanation\n- **Features (X):** Sepal length, sepal width, petal length, and petal width. These are the model inputs.\n- **Target (y):** `species`, the flower category the model would predict.\n- The target is kept separate from X so the model does not receive the answer as an input.\n\nExpected shapes: `X = (150, 4)` and `y = (150,)`.
